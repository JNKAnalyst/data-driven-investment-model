# 04 — Model Comparison


Side-by-side comparison of the three models using cross-validation.

Numbers below are produced by the most recent run of `src.models.run` against the dataset configured in `config/config.yaml`.


```python
import sys; sys.path.append('..')
import pandas as pd
from sklearn.model_selection import train_test_split
from src.data import load_config, load_dataset, split_features_target
from src.features import engineer_features
from src.models import build_models
from src.evaluate import cross_validate_model, format_results_table

cfg = load_config('../config/config.yaml')
df = engineer_features(load_dataset('../' + cfg['data']['data_path']))
X, y = split_features_target(df, cfg['data']['target_column'])
X_train, _, y_train, _ = train_test_split(X, y, test_size=cfg['data']['test_size'], random_state=cfg['data']['random_state'])

scores = [cross_validate_model(name, est, X_train, y_train, cv_folds=cfg['evaluation']['cv_folds'], random_state=cfg['data']['random_state']) for name, est in build_models(cfg['models']).items()]
print(format_results_table(scores))
```
